# Merge TTL Files and Visualize with RDF Grapher
This notebook merges two RDF Turtle files from GitHub and visualizes the merged graph using the RDF Grapher service.

```python
pip install rdflib
```

```python
from rdflib import Graph
from IPython.display import HTML, display
import urllib.parse
```

## Step 1: Load and merge the RDF graphs

```python
from rdflib import Graph

g = Graph()

g.parse("data/darkside.ttl", format="turtle")
g.parse("data/wall.ttl", format="turtle")

print(f"Merged graph contains {len(g)} triples")
```

## Step 2: Save merged graph locally

```python
g.serialize('merged.ttl', format='turtle')
print('Saved merged.ttl')
```

## Step 3: Generate RDF Grapher URL

```python
rdf_text = g.serialize(format='turtle')
encoded_rdf = urllib.parse.quote(rdf_text)

graph_url = (
    'https://www.ldf.fi/service/rdf-grapher'
    f'?rdf={encoded_rdf}&from=ttl&to=svg'
)

from IPython.display import HTML, display

display(
    HTML(
        f"""
        <p>
          {graph_url}
            🔍 Open graph in a separate window
          </a>
        </p>
        """
    )
)
```
